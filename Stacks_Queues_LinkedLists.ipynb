````python
# Node

class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
````

# 1. Stacks

````python
# Stack

# methods: push, pop, peek, is_empty

class Stack:
    def __init__(self):                  
        self.top = None
        self.size = 0

    def push(self, value):
        new_node = Node(value)
        new_node.next = self.top
        self.top = new_node
        self.size += 1
        
    def pop(self):
        if self.is_empty():
            raise IndexError('Stack is empty')  
        value = self.top.value
        self.top = self.top.next
        self.size -= 1
        return value

    def peek(self):
        if self.is_empty():
            raise IndexError('Stack is empty')
        return self.top.value

    def is_empty(self):
        return self.top is None
````

````python
s = Stack()

s.push(5)
s.push(7)
print(s.peek())
print(s.size)
print(s.pop())
print(s.top.value)
print(s.size)
````

# 2. Queue

````python
# Queue

# methods: enqueue, dequeue, peek, isEmpty, size

class Queue:
    
    def __init__(self):
        self.front = None
        self.rear = None
        self.size = 0

    def enqueue(self, value):
        new_node = Node(value)
        if self.isEmpty():
            self.front = new_node
            self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node
        self.size += 1

    def dequeue(self):
        if self.isEmpty():
            raise IndexError('Queue is empty')
        value = self.front.value
        self.front = self.front.next
        if self.front is None:
            self.rear = None
        self.size -= 1
        return value

    def peek(self):
        if self.isEmpty():
            raise IndexError('Queue is empty')
        return self.front.value

    def isEmpty(self):
        return self.front is None

    def __len__(self):
        return self.size

````

````python
q = Queue()

q.enqueue(5)
q.enqueue(7)
q.enqueue(9)
print(len(q))
print(q.dequeue())
print(q.peek())
````

# 3. Linked List

A Linked List is a linear data structure where elements (nodes) are stored non-contiguously. Each node contains Data (value) and Referene (pointer) to the next node (and optionally previous node)

````python
class LinkedList:

    def __init__(self):
        self.head = None
        self.size = 0

    def append(self, value):
        new_node = Node(value)
        if self.head is None:
            self.head = new_node
        else:
            current = self.head
            while current.next is not None:
                current = current.next
            current.next = new_node
        self.size += 1

    def prepend(self, value):
        new_node = Node(value)
        new_node.next = self.head
        self.head = new_node
        self.size += 1
        

    def insert_after(self, target_value, new_value):
        current = self.head
        while current is not None:
            if current.value == target_value:
                new_node = Node(new_value)
                new_node.next = current.next
                current.next = new_node
                break
            current = current.next
        raise IndexError('target_value is not found')         
            

    def delete(self, value):
        if self.head is None:
            raise IndexError("Empty List")

        # Handle head deletion
        if self.head.value == value:
            self.head = self.head.next
            self.size -= 1
            return True

        # Traverse
        curr = self.head
        while curr.next is not None:
            if curr.next.value == value:
                curr.next = curr.next.next
                self.size -= 1
                return True
            curr = curr.next

        raise IndexError('Value not found')
        

    def search(self, value):
        current = self.head
        node_num = 1
        while current is not None:
            if current.value == value:
                return "Value found at node: " + str(node_num)
            node_num += 1
            current = current.next
        return "Value not found"
        

    def reversed(self):
        prev = None
        curr = self.head
        while curr is not None:
            next_ = curr.next
            curr.next = prev
            prev = curr
            curr = next_
        self.head = prev
   

    def display(self):
        list_ = []
        if self.size == 0:
            return 'Linked List is empty'
        curr = self.head
        while curr is not None:
            list_.append(curr.value)
            curr = curr.next
        return list_
            

    def __len__(self):
        return self.size
````

**1. Valid Parenthesis Problem**

Given a string s containing just the characters '(', ')', '{', '}', '[' and ']', determine if the input string is valid.

A string is valid if:

- Open brackets must be closed by the same type of brackets.

- Open brackets must be closed in the correct order.

- Every closing bracket has a corresponding open bracket of the same type.

````python
def validParenthesis(s):
    map_ = { '(' : ')', '{' : '}', '[' : ']' }
    stk = Stack()
    for i in s:
        if i in map_.values():
            if stk.is_empty():
                return False
            if stk.peek() != i:
                return False
            else:
                stk.pop()
        elif i in map_.keys():
            stk.push(map_[i])
    if stk.is_empty():
        return True
````

````python
s1 = '(}}})]'
s2 = '  '
s3 = '([)]'
s4 = '([])'

print(validParenthesis(s1))
print(validParenthesis(s2))
print(validParenthesis(s3))
print(validParenthesis(s4))
````

**2. Min Stack**

Design a stack that supports the following operations in O(1) time:

push(val) – push element val onto the stack.

pop() – remove the element on top of the stack.

top() – get the top element.

getMin() – retrieve the minimum element in the stack.

You may use any built-in data structures (like Python list) internally, but all operations must be O(1).

```
minStack = MinStack()
minStack.push(-2)
minStack.push(0)
minStack.push(-3)
minStack.getMin()   --> returns -3
minStack.pop()
minStack.top()      --> returns 0
minStack.getMin()   --> returns -2
```

````python
class MinStack(list):
    def __init__(self):
        super().__init__()

    def push(self, value):
        min_val = value if self.getMin() is None else min(value, self.getMin())
        self.append((value, min_val))
       
    def pop(self):
        return super().pop()[0]

    def top(self):
        return self[-1][0]

    def getMin(self):
        return None if len(self) == 0 else self[-1][1]
    
````

````python
minStack = MinStack()
minStack.push(-2)
minStack.push(0)
minStack.push(-3)
print(minStack.getMin()) 
print(minStack.pop())
print(minStack.top()) 
print(minStack.getMin())   
````

**3. Reverse Polish Notation**

Reverse Polish Notation (RPN) is a mathematical notation where operators follow their operands. For example, "3 4 +" means 3 + 4 = 7.

You are given an array of strings tokens that represents an arithmetic expression in RPN.

Evaluate the expression and return the integer result.

Rules
Valid operators: '+', '-', '*', '/'.

Each operand may be an integer or another expression result.

Division between two integers should truncate toward zero (floor division for positive/negative).

The given RPN expression is always valid (no division by zero, enough operands, etc.).

The result and all intermediate values fit in 32-bit signed integer range.

```
Examples
Input	Output	Explanation
["2","1","+","3","*"]	9	(2 + 1) * 3 = 9
["4","13","5","/","+"]	6	4 + (13 / 5) = 4 + 2 = 6
["10","6","9","3","+","-11","*","/","*","17","+","5","+"]	22	((10 * (6 / ((9 + 3) * -11))) + 17) + 5 = 22
["3","4","+"]	7	3 + 4 = 7
["15","7","1","1","+","-","/","3","*","2","1","1","+","+","-"]	5	(complex expression yields 5)
```

Constraints
1 <= tokens.length <= 10^4

tokens[i] is either an operator (+, -, *, /) or an integer (possibly negative).

Valid RPN expression.

````python
def evalRPN(tokens: list[str]) -> int:
    stk = Stack()
    for i in tokens:
        if i in ['+', '*', '-', '/']:
            num2 = stk.pop()
            num1 = stk.pop()
            if i == '+':
                result = num1 + num2
            elif i == '-':
                result = num1 - num2
            elif i == '*':
                result = num1 * num2
            else:
                result = int(num1 / num2)
            stk.push(result)
        else:
            stk.push(int(i))
    return stk.pop()
            
````

````python
tokens1 = ["2","1","+","3","*"]
tokens2 = ["4","13","5","/","+"]
tokens3 = ["15","7","1","1","+","-","/","3","*","2","1","1","+","+","-"]

print(evalRPN(tokens1))
print(evalRPN(tokens2))
print(evalRPN(tokens3))
````

**4. Linked List Cycle Detection**

You are given the head of a singly linked list. The list is made of ListNode objects:

The list may be empty (head == None) or contain one or more nodes. Each node has a next pointer to another node or None.

Task
Detect whether the linked list has a cycle.
A cycle exists if, by following the next pointers repeatedly, you eventually come back to a node you have visited before. In other words, there is no node with next == None; the list never ends.

Return True if there is a cycle, otherwise return False.


````python
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None


# node1 = ListNode(1)
# node2 = ListNode(2)
# node3 = ListNode(3)
# node1.next = node2
# node2.next = node3
# head = node1


node1 = ListNode(1)
node2 = ListNode(2)
node3 = ListNode(3)
node1.next = node2
node2.next = node3
node3.next = node2 
head = node1
````

````python
def hasCycle(head):
    map_ = {}
    curr = head
    while curr is not None:
        map_[curr] = curr.next
        if curr.next in map_:
            return True
        curr = curr.next
    return False


# O(N) time, O(N) space
````
