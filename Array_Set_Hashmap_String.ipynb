### 1. Warm-up Problems

**1.1. Contains Duplicate**

Given an integer array nums, return True if any value appears at least twice in the array, and False if every element is distinct.

Example 1:

```
Input: nums = [1,2,3,1]
Output: True
```
Example 2:

```
Input: nums = [1,2,3,4]
Output: False
```
Example 3:

```
Input: nums = [1,1,1,3,3,4,3,2,4,2]
Output: True
```




````python
def containsDuplicate(nums):
    seen = {}
    for i in nums:
        if i in seen:
            return True
        seen[i] = 1
    return False
````

````python
nums = [1,2,3,1]

containsDuplicate(nums)
````

**1.2. Missing Number**

Given an array nums containing n distinct numbers in the range [0, n], return the only number in the range that is missing from the array.

Example 1:

```
Input: nums = [3,0,1]
Output: 2
Explanation: n = 3, numbers in range [0,3] are 0,1,2,3. 2 is missing.
```

Example 2:

```
Input: nums = [0,1]
Output: 2
Explanation: n = 2, range [0,2] → missing 2.
```

Example 3:

```
Input: nums = [9,6,4,2,3,5,7,0,1]
Output: 8
```



````python
def missingNumber(nums: list[int]) -> int:
    n = len(nums)
    expected_sum = int(n*(n+1)/2)
    actual_sum = sum(nums)
    return expected_sum - actual_sum
        
````

````python
nums = [9,6,4,2,3,5,7,0,1]

missingNumber(nums)
````

**1.3. Intersection of Two Arrays**

Given two integer arrays nums1 and nums2, return an array of their intersection (unique elements that appear in both arrays). The result can be in any order.

Example 1:

```
Input: nums1 = [1,2,2,1], nums2 = [2,2]
Output: [2]
```
Example 2:

```
Input: nums1 = [4,9,5], nums2 = [9,4,9,8,4]
Output: [4,9]   (order may vary)
```


````python
def intersection(nums1: list[int], nums2: list[int]) -> list[int]:
    nums1 = set(nums1)
    nums2 = set(nums2)
    return nums1 & nums2
````

````python
nums1 = [4,9,5]
nums2 = [9,4,9,8,4]

intersection(nums1, nums2)
````

**1.4. Majority Element**

Given an array nums of size ```n```, return the majority element. The majority element is the element that appears more than ```⌊n / 2⌋``` times. You may assume that the majority element always exists in the array.

Example 1:
```
Input: nums = [3,2,3]
Output: 3
```
Example 2:
```
Input: nums = [2,2,1,1,1,2,2]
Output: 2
```
Constraints:
- n == nums.length
- 1 <= n <= 5 * 10^4
- -10^9 <= nums[i] <= 10^9

**Follow‑up:** Can you solve it in O(n) time and O(1) space? (Hint: Boyer‑Moore Voting Algorithm)

````python
from collections import Counter

def majorityElement(nums: list[int]) -> int:
    thresh = len(nums)//2
    counter = Counter(nums)
    major_elem, major_count = counter.most_common(1)[0]
    if major_count > thresh:
        return major_elem
    
````

````python
nums = [2,2,1,1,1,2,2]
majorityElement(nums)
````

````python
# Recommended Method: Boyer-Moore Voting Algorithm. (constant time)

def majorityElement(nums: list[int]) -> int:
    candidate = None
    count = 0
    
    for i in nums:
        if count == 0: candidate = i
            
        if i == candidate: count += 1
        else: count -= 1

    return candidate
        
````

**1.5. Two Sum**

Given an array of integers `nums` and an integer `target`, return the indices of the two numbers in that array that add up to `target`.

Example:

```
Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: nums[0] + nums[1] = 2 + 7 = 9, so we return [0, 1].
```


````python
def twoSum(nums, target):
    seen = {}
    for idx, val in enumerate(nums):
        if target - val in seen:
            return [seen[target - val], idx] 
        seen[val] = idx

````

````python
nums = [2,7,11,15]
target = 9
twoSum(nums, target)
````

### 2. Sliding Window Problems

Involves contiguous subarrays or substrings

Maintains a window that "slides" over the input

Available in both fixed and variable sizes

You maintain two pointers - left and right that define the current window. you expand the right pointer to include new elements and shrink from the right when the window becomes invalid.

- right pointer moves forward one step at a time
- left pointer when needed to maintain window constraints

This makes every element enter and exit the window once giveing time complexity 0(N)





**2.1. Max sum of subarray of size k**

Fixed Window Length

````python
def max_sum_subarray_k(arr, k):
    if not arr or k <= 0 or k > len(arr):
        return None
    left = 0
    window_sum = sum(arr[:k])
    max_sum = float('-inf')
    while left + k < len(arr):
        window_sum += arr[left+k] - arr[left]
        max_sum = max(max_sum, window_sum)
        left += 1
    return max_sum
````

````python
arr = [1,8,7,3,5,1,9,5,8,2,6]
k = 2
max_sum_subarray_k(arr, k)
````

**2.2. Longest sub-string without repeating characters**

Variable Window Length

````python
def LSWRC(s: str) -> int:
    left = 0
    chr_idx = {}
    max_len = 0
    
    for right, chr_ in enumerate(s): # move right one step at a time
        if chr_ in chr_idx and chr_idx[chr_] >= left:  # adjust left if unique condition is invalidated
            left = chr_idx[chr_] + 1
        chr_idx[chr_] = right # update character index dictionry
        max_len = max(max_len, right - left + 1)
    
    return max_len
````

````python
s = "abcabcbb"
LSWRC(s)
````

Given an array of integers nums and an integer k, return the total number of subarrays whose sum equals k.

A subarray is a contiguous, non-empty sequence of elements within the array.


Example 1:

```
Input: nums = [1,1,1], k = 2
Output: 2
Explanation: The subarrays [1,1] (indices 0-1) and [1,1] (indices 1-2) both sum to 2.
```

Example 2:

```
Input: nums = [1,2,3], k = 3
Output: 2
Explanation: Subarrays [1,2] (sum 3) and [3] (sum 3) both sum to 3.
```

Example 3:

```
text
Input: nums = [1,-1,0], k = 0
Output: 3
Explanation: Subarrays [1,-1], [0], and [1,-1,0] all sum to 0.
```

````python
def subArraySum2(nums, k):
    prefix = 0
    seen = {0:1}
    count = 0
    for i in nums:
        prefix += i
        count += seen.get(prefix - k, 0)
        seen[prefix] = seen.get(prefix, 0) + 1
    return count
````

````python
nums = [1,2,3]
k = 3
subArraySum2(nums, k)
````
