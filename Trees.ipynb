# Trees

Why Tree? 

Trees provides a non-linear, hierarchical framework that dramatically optimises (O(logn) time) data organisation, searching, processing compared to linear structures like array, matrix (O(n) time).

UseCases: 
- Hierarchical Data Representation (File Systems, Web Browsers DOM, BI Charts)
- High-Performance Database Indexing (B-Trees, LSM Trees)
- Text Processing and Search Features (Tries or Prefix Trees, Suffix Trees)
- Compilers and Code Parsing (Abstract Syntax Trees)
- Network Routing Optimisation (Routig Tables, Spanning Trees)
- Machine Learning and Gaming (Decision Trees, Game Trees)
- Core Memory and Library Performance (Heaps, AVL trees, Merkle Trees)

````python
from collections import deque


class TreeNode:
    
    # Constructor
    def __init__(self, value, left = None, right = None):
        self.value = value
        self.left = left
        self.right = right

    ############################################################### Traversal  ###################################################################

    ########################################################### Depth-First Search  ##############################################################
    
    # InOrder - recursive
    def inorder(self, result = None):
        # initialise result
        if result is None:
            result = []
        # store left sub-tree values
        if self.left:
            self.left.inorder(result)               # recursion
        # store value
        result.append(self.value)
        # store right sub-tree values
        if self.right:
            self.right.inorder(result)
        return result

    # Inorder - iterative
    
    # what i wrote:
    # def inorder_itr(self):
    #     result = [self.value]
    #     stack = [self]
    #     while stack:
    #         node = stack.pop()
    #         # print(node.value)
    #         sub_result = [node.value]
    #         if node.right:
    #             stack.append(node.right)
    #             sub_result += [node.right.value]
    #         if node.left:
    #             stack.append(node.left)
    #             sub_result = [node.left.value] + sub_result
            
    #         for ix, val in enumerate(result):
    #             if val == node.value:
    #                 result = result[:ix] + sub_result + result[ix+1:]
    #     return result
                
    # official
    def inorder_itr(self):
        result = []
        stack = []
        node = self
        while stack or node:
            while node:
                stack.append(node)     
                node = node.left

            node = stack.pop()
            result.append(node.value)

            node = node.right
        return result
        

    # PreOrder (middle - left - right)
    def preorder(self, result = None):
        # initalise result
        if result is None:
            result = []
        # store self node value
        result.append(self.value)
        # store left sub-tree values
        if self.left:
            self.left.preorder(result)
        # store right sub-tree values
        if self.right:
            self.right.preorder(result)
        return result

    # PreOrder - iterative
    def preorder_itr(self):
        result = []
        stack = [self]
        while stack:
            node = stack.pop()
            result.append(node.value)
            if node.right:
                stack.append(node.right)
            if node.left:
                stack.append(node.left)
        return result


    # PostOrder (left - right - middle)
    def postorder(self, result = None):
        # initialise result
        if result is None:
            result = []
        # store left sub-tree values
        if self.left:
            self.left.postorder(result)
        # store right sub-tree values
        if self.right:
            self.right.postorder(result)
        # store self node value
        result.append(self.value)
        return result

    # PostOrder - iterative
    def postorder_itr(self):
        result = []
        stack1 = [self]
        stack2 = []
        while stack1:
            node = stack1.pop()
            stack2.append(node)
            if node.left:
                stack1.append(node.left)
            if node.right:
                stack1.append(node.right)
        while stack2:
            node = stack2.pop()
            result.append(node.value)
        return result


    ############################################################ Breadth-First Search ##########################################################

    # level Order treaversal
    def levelOrder(self):
        result = []
        queue = deque([[self]])
        while queue:
            nodes = queue.popleft()
            r_ = []
            q_ = []
            for node in nodes:
                r_.append(node.value)
                if node.left:
                    q_.append(node.left)
                if node.right:
                    q_.append(node.right)
            result.append(r_)
            if q_:
                queue.append(q_)
        return result


#####################################################################################################################################################

# tree object
tree = TreeNode(7, TreeNode(2, TreeNode(1), TreeNode(5)), TreeNode(9, TreeNode(8)))
````

````python
class BST(TreeNode):

    ######################################################## Binary Search Tree ##########################################################

    # validate BST
    # one way O(N) space
    def isValidBST(self):
        result = self.inorder()
        return all([x < y for x, y in zip(result[:-1], result[1:])])

    # Recursive range bound O(H) space where H is height of tree
    def isValidBST2(self, min_val=float('-inf'), max_val=float('inf')):
        # Check current node's value is within bounds
        if not (min_val < self.value < max_val):
            return False
        
        left_valid = True
        right_valid = True
        
        if self.left:
            left_valid = self.left.isValidBST2(min_val, self.value)
        if self.right:
            right_valid = self.right.isValidBST2(self.value, max_val)
        
        return left_valid and right_valid

    # Iterative inorder
    def isValidBST3(self):
        prev = None
        curr = self
        stack = []
        while stack or curr:
            # go as left as possible
            while curr:
                stack.append(curr)
                curr = curr.left
            # pop and process    
            curr = stack.pop()
            #check order
            if prev is not None and curr.value <= prev:
                return False
            prev = curr.value
            # move to right
            curr = curr.right
        return True
        

    # insert into
    def insertIntoBST(self, val):
        if val > self.value:
            if self.right:
                self.right.insertIntoBST(val)
            else:
                self.right = TreeNode(val)
        elif val < self.value:
            if self.left:
                self.left.insertIntoBST(val)
            else:
                self.left = TreeNode(val)
        return self

        
````

**Problem 1: Max Depth**

Given the root of a binary tree, return its maximum depth.

Maximum depth is the number of nodes along the longest path from the root down to the farthest leaf node. i.e. number of levels in the tree

````python
def maxDepth(root):
    depth = 0
    if root is None:
        return depth
    queue = deque([root])
    while queue:
        depth += 1
        for _ in range(len(queue)):
            node = queue.popleft()
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
    return depth
````

**Problem 2: Invert Tree**

Given the root of a binary tree, invert the tree (mirror it) and return the root

````python
# inverts in place
def invertBT(root):
    if root:
        root.left, root.right = root.right, root.left
        invertBT(root.left)
        invertBT(root.right)
````

````python
tree = TreeNode(7, TreeNode(2, TreeNode(1), TreeNode(5)), TreeNode(9, TreeNode(8)))

print(tree.inorder())

invertBT(tree)

print(tree.inorder())
````

````python
# does not invert inplace, return a new inverted tree node 

def invertBT2(root):
    if root is None:
        return None
    new_node = TreeNode(root.value)
    new_node.left = invertBT2(root.right)
    new_node.right = invertBT2(root.left)
    return new_node
````

````python
tree = TreeNode(7, TreeNode(2, TreeNode(1), TreeNode(5)), TreeNode(9, TreeNode(8)))

print(tree.inorder())

new_tree = invertBT2(tree)

print(tree.inorder())
print(new_tree.inorder())
````

**Problem 3: Same Tree**

Given the roots of two binary trees p and q, write a function to check if they are the same tree

````python
def isSameTree(p, q):
    if not p and not q:
        return True
    if not p or not q :
        return False
    return (p.value == q.value) and isSameTree(p.left, q.left) and isSameTree(p.right, q.right)
````

````python
tree1 = TreeNode(7, TreeNode(2, TreeNode(1), TreeNode(5)), TreeNode(9, TreeNode(8)))

tree2 = TreeNode(7, TreeNode(2, TreeNode(1), TreeNode(5)), TreeNode(9, TreeNode(8)))

isSameTree(tree1, tree2)
````

**Problem 4: Lowest Common Ancestor**

````python
# LCA: attempt 1


def path_exists(node1, node2):
    if node1 is None:
        return False
    if node1.value == node2.value:
        return True
    return path_exists(node1.left, node2) or path_exists(node1.right, node2)

def lca(tree, node1, node2):
    stack = [tree]
    lca = None
    while stack:
        curr = stack.pop()
        if path_exists(curr, node1) and path_exists(curr, node2):
            lca = curr
        if curr.left:
            stack.append(curr.left)
        if curr.right:
            stack.append(curr.right)
    return lca
````

````python
# LCA - official solution

def lca(root, p, q):
    # base case
    if root is None or root is p or root is q:
        return root

    # recurse left and right
    left = lca(root.left, p, q)
    right = lca(root.right, p, q)

    # If both sides returned a node, current root is LCA
    if left and right:
        return root
    
    # if both sides returned a node, current root is laca
    return left if left else right
````

**Problem 5: Path Sum**

Given a binary and a target sum, check if there exists a root-to-leaf path where the sum of node values equals the target

````python
def hasPathSum(root, targetSum):
    if root is None:
        if targetSum == 0: return True
        else: return False
    remaining = targetSum - root.value
    left = hasPathSum(root.left, remaining)
    right = hasPathSum(root.right, remaining)
    return left or right
````
