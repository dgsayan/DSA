### 1. LRU Cache

```python
# LRU using OrderedDict 

from collections import OrderedDict

class LRU:
    
    def __init__(self, len_):
        self.dict_ = OrderedDict()
        self.len_ = len_

    def get(self, key):
        val = None
        if key in self.dict_:
            val = self.dict_[key]
            self.dict_.move_to_end(key)
        return val

    def put(self, key, value):
        if (len(self.dict_) == self.len_) and (key not in self.dict_):
            self.dict_.popitem(last = False)
        self.dict_[key] = value
        self.dict_.move_to_end(key)
        
```

```python
lru = LRU(3)
lru.put('a', 5)
lru.put('b', 2)
lru.put('c', 1)
print(lru.get('a'))
lru.put('d', 1)
print(lru.get('a'))
print(lru.get('b'))
```

```python
# LRU using hash map + linked list

class Node:
    def __init__(self, key = 0, val = 0):
        self.key = key
        self.val = val
        self.prev_ = None
        self.next_ = None


class LRUCache:
    def __init__(self, capacity):
        self.capacity = capacity
        self.cache = {}
        self.head = Node()
        self.tail = Node()
        self.head.next = self.tail
        self.tail.prev = self.head

    def _add_to_tail(self, node):
        node.prev = self.tail.prev 
        node.next = self.tail
        self.tail.prev.next = node
        self.tail.prev = node
        
    def _remove(self, node):
        node.prev.next = node.next
        node.next.prev = node.prev

    def get(self, key):
        if key not in self.cache:
            return -1
        node = self.cache[key]
        self._remove(node)
        self._add_to_tail(node)
        return node.val

    def put(self, key, val):
        if key in self.cache:
            node = self.cache[key]
            node.val = val
            self._remove(node)
            self._add_to_tail(node)
        if len(self.cache) == self.capacity:
            least_recent = self.head.next
            self._remove(least_recent)
            del self.cache[least_recent.key]
        new_node = Node(key, val)
        self._add_to_tail(new_node)
        self.cache[key] = new_node
    
```

```python
cache = LRUCache(2)
cache.put(1, 1)
cache.put(2, 2)
print(cache.get(1))    # 1
cache.put(3, 3)        # evicts key 2
print(cache.get(2))    # -1
cache.put(4, 4)        # evicts key 1
print(cache.get(1))    # -1
print(cache.get(3))    # 3
print(cache.get(4))    # 4
```

```python
# trie official implementation

class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word: str) -> None:
        node = self.root
        for ch in word:
            if ch not in node.children:
                node.children[ch] = TrieNode()
            node = node.children[ch]
        node.is_end = True

    def search(self, word: str) -> bool:
        node = self.root
        for ch in word:
            if ch not in node.children:
                return False
            node = node.children[ch]
        return node.is_end

    def startsWith(self, prefix: str) -> bool:
        node = self.root
        for ch in prefix:
            if ch not in node.children:
                return False
            node = node.children[ch]
        return True
```

```python
trie = Trie()
trie.insert("apple")
print(trie.search("apple"))   # True
print(trie.search("app"))     # False
print(trie.startsWith("app")) # True
trie.insert("app")
print(trie.search("app"))     # True
print(trie.startsWith("ap"))  # True
print(trie.startsWith("b"))   # False
```

### 4. Serialise - Deserialise a Binary Tree

```python
class TreeNode:
    def __init__(self, val, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def serialise_preorder(root: Optional[TreeNode]) -> str:
    result = []
    if root is None:
        return ""
    stack = [root]
    while stack:
        node = stack.pop()
        if node:
            result.append(str(node.val))
            if node.right or node.left:
                stack.append(node.right)
                stack.append(node.left)
        else:
            result.append("Null")

    return ','.join(result)
```

```python
tree = TreeNode(5, TreeNode(2, TreeNode(3)), TreeNode(7, right = TreeNode(8)))
```

```python
class TreeNode:
    def __init__(self, val, left = None, right = None, left_visited = 0, right_visited = 0, prev = None, depth = 1):
        self.val = val
        self.left = left
        self.right = right
        self.left_visited = left_visited
        self.right_visited = right_visited
        self.prev = prev
        self.depth = depth
```

```python
from math import log2

def deserialise_preorder(serial: str) -> Optional[TreeNode]:

    if serial == '': return None
    
    serial = [None if i == 'Null' else int(i) for i in serial.split(',')]

    serial = serial[::-1]
    
    max_depth = int(log2(len(serial))) + 1
    
    node = None

    while serial:
        
        val = serial.pop()

        if node is None:
            node = TreeNode(val)

        elif node.left_visited == 0 and node.depth < max_depth:
            node.left_visited = 1 
            if val is not None:
                node.left = TreeNode(val, prev = node, depth = node.depth + 1)
                node = node.left

        elif node.left_visited == 1 and node.depth < max_depth:
            node.right_visited = 1
            if val is not None:
                node.right = TreeNode(val, prev = node, depth = node.depth + 1)
                node = node.right

        elif (node.depth == max_depth) or (node.left_visited == 1 and node.right_visited == 1):
            while node.right_visited == 1:
                node = node.prev
            node.right_visited = 1
            if val is not None:
                node.right = TreeNode(val, prev = node, depth = node.depth + 1)
                node = node.right
            

    while node.prev is not None:
        node = node.prev

    return node
    
     
```
